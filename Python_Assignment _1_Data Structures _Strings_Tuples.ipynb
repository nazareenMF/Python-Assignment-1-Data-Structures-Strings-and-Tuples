### 2. String Slicing and Indexing:

### Write a Python program using the above concatenated string as input and performs the following tasks:

```python
# a. Print the first character of the string

Result2=Result1+String3                      
print(Result2)
print("First Character:",Result2[0])   # Indexing is done to find first character
```

```python
# b. Print the last character of the string

Result2=Result1+String3                      
print(Result2)
print("Last Character:",Result2[-1])     #indexing to find last position
```

```python
# c. Print the first 5 characters of the string.

Result2=Result1+String3                      
print(Result2)
print("First Five Characters:",Result2[:5])  # Slicing first 5 characters where first index is 0
```

```python
# d. Print the last 11 characters of the string.

Result2=Result1+String3                      
print(Result2)
print("Last eleven characters:",Result2[-11:])  # Slicing last 11 characters where last index is -1
```

```python
# e. Print the string in reverse.

Result2=Result1+String3                      
print(Result2)
print("Reversed String:",Result2[::-1])   # result[::-1] reverses the entire string
```

```python
# f. Use slicing and print the word “Python” from the existing string.

Result2=Result1+String3                      
print(Result2)
print("The word Python:",Result2[23:29])
```

## 3. String Methods:

### Write a Python program that takes a string,strM = “Python beginner tutorial” and perform the following tasks:

```python
# a. Convert the sentence to uppercase

strM ="Python beginner tutorial"  #assigning value to strM
print(strM)
print(strM.upper())    # Converting string to uppercase
```

```python
# b. Convert the sentence to lowercase.

print(strM)
print(strM.lower())    # Converting string to lowercase
```

```python
# c. Use Capitalize and return the sentence to the original input form.

print(strM.capitalize())  # capitalize() capitalizes the first character and converts the rest to lowercase.
print(strM)        #The original string remains unchanged because Python strings are immutable.
```

```python
# d. Count the total number of occurrences of character ‘t’ in the string.

print(strM)
print("No:of occurences of't':",strM.count('t'))
```

```python
''' e. Replace all occurrences of “Python” with “Machine Learning” 
in the input string strM = “Python beginner tutorial” '''

print("Original String:",strM)

# replacing Python--> Machine learning

print("Changed string:",strM.replace('Python','Machine Learning'))
```

## Tuples (Creation, Modification and Access) :

### Create 1st tuple with values -> (10, 20, 30), 2nd tuple with values -> (40, 50, 60):

```python
# a. Concatenate the two tuples and store it in “t_combine”

tuple1=(10,20,30)   #Creating first tuple
tuple2=(40,50,60)   #Creating second tuple

t_combine= tuple1 + tuple2    #Concatenating the two tuples and storing it in t_combine

print("First tuple:", tuple1)
print("Second tuple:", tuple2)
print("Concatenated tuple:",t_combine)
```

```python
# b. Repeat the elements of “t_combine” 3 times

print (t_combine)     
Repeated_tuple= t_combine*3   #Repeating the elements of t_combine 3 times
print("Repeated tuple:", Repeated_tuple)  
```

```python
#  c. Access the 3rd element from “t_combine”

print (t_combine) 
print("Third element of t_combine:",t_combine[2])   #Indexing 
```

```python
#  d. Access the first three elements from “t_combine”

print (t_combine) 
print("First 3 elements of t_combine:",t_combine[:3])  
```

```python
# e. Access the last three elements from “t_combine”

print (t_combine)
print("Last 3 elements of t_combine:",t_combine[-3:])
```
